# SQL Database Setup

## Objective

This notebook demonstrates how the cleaned marketing datasets are integrated into a PostgreSQL database.

The primary goals are:

- Establish a connection to PostgreSQL
- Create database tables
- Import cleaned datasets
- Validate successful data insertion
- Prepare the data for SQL-based analysis and reporting

```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
ad_df = pd.read_csv("ad_spend_data.csv")

web_df = pd.read_csv("web_analytics_log.csv")

crm_df = pd.read_csv("crm_conversion_data.csv")
```

```python
# Create a SQLAlchemy engine to connect to the PostgreSQL database
engine = create_engine(
    "postgresql://postgres:SQLogin1@localhost:5432/infotact"
)
```

```python
# Create the ad_spend_data table in PostgreSQL
ad_df.to_sql(
    "ad_spend",
    engine,
    if_exists="replace",
    index=False
)
```

```python
# Create the web_analytics_log table in PostgreSQL
web_df.to_sql(
    "web_analytics",
    engine,
    if_exists="replace",
    index=False
)
```

```python
# Create the crm_conversions table in PostgreSQL
crm_df.to_sql(
    "crm_conversions",
    engine,
    if_exists="replace",
    index=False
)
```

```python
pd.read_sql(
    "SELECT * FROM ad_spend LIMIT 5",
    engine
)
```

## Database Validation

The following queries verify that all datasets have been successfully imported into PostgreSQL.

```python
# Verify row counts

queries = {
    "Ad Spend": "SELECT COUNT(*) FROM ad_spend;",
    "Web Analytics": "SELECT COUNT(*) FROM web_analytics;",
    "CRM Conversions": "SELECT COUNT(*) FROM crm_conversions;"
}

for table, query in queries.items():
    result = pd.read_sql(query, engine)
    print(f"{table}:")
    print(result)
    print("-" * 40)
```
