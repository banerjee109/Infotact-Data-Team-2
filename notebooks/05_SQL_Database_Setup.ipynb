```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
ad_df = pd.read_csv("ad_spend_data.csv")

web_df = pd.read_csv("web_analytics_log.csv")

crm_df = pd.read_csv("crm_conversion_data.csv")
```

```python
engine = create_engine(
    "postgresql://postgres:SQLogin1@localhost:5432/infotact"
)
```

```python
ad_df.to_sql(
    "ad_spend",
    engine,
    if_exists="replace",
    index=False
)
```

```python
web_df.to_sql(
    "web_analytics",
    engine,
    if_exists="replace",
    index=False
)
```

```python
crm_df.to_sql(
    "crm_conversions",
    engine,
    if_exists="replace",
    index=False
)
```

```python
pd.read_sql(
    "SELECT * FROM ad_spend LIMIT 5",
    engine
)
```
